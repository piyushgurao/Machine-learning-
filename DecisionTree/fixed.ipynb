## Decision tree classifier

```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
from sklearn.datasets import load_iris
iris=load_iris()
```

```python
iris
```

```python
print(iris.target)
```

```python
iris.data
```

```python
X=pd.DataFrame(iris.data,columns=iris.feature_names)
```

```python
X.head()
```

```python
len(X)
X.shape
```

```python
y=pd.DataFrame(iris.target,columns=['target'])
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=10)
```

```python
X_train.shape
```

## apply decision tree classifier

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
treeclassifier=DecisionTreeClassifier()
```

```python
treeclassifier.fit(X_train,y_train)
```

```python
## visulize the decision tree
from sklearn import tree
plt.figure(figsize=(15,10))
tree.plot_tree(treeclassifier,filled=True)
```

- we can observe from above model is overfitted 
- there is no need for further splitting after depth 2 

```python
treeclassifier2=DecisionTreeClassifier(max_depth=2)
```

```python
treeclassifier2.fit(X_train,y_train)
```

```python
## visulize the decision tree
plt.figure(figsize=(15,10))
tree.plot_tree(treeclassifier2,filled=True)
```

```python
y_pred=treeclassifier2.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix,classification_report,accuracy_score
```

```python
print("accuracy score: ", accuracy_score(y_test,y_pred))
print("confusion matrics : \n",confusion_matrix(y_test,y_pred))
print("classification_report :\n",classification_report(y_test,y_pred))
```

# Decision tree classifier preprunning and hyperparameter tuning

```python
params={
    'criterion':['gini', 'entropy', 'log_loss'],
    'splitter':['best','random'],
    'max_depth':[1,2,3,4,5],
    'max_features':['auto','sqrt','log2']
}
```

```python
params
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
treemodel=DecisionTreeClassifier()
```

```python
grid=GridSearchCV(treemodel,param_grid=params,cv=5,scoring='accuracy',refit=True)
```

```python
import warnings
warnings.filterwarnings('ignore')
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
score=grid.score(X_test,y_test)
score
```

```python
y_pred=grid.predict(X_test)
print("accuracy score: ", accuracy_score(y_test,y_pred))
print("confusion matrics : \n",confusion_matrix(y_test,y_pred))
print("classification_report :\n",classification_report(y_test,y_pred))
```
