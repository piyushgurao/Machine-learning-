```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## lets create some synthetic data points

```python
from sklearn.datasets import make_classification
```

```python
X,y=make_classification(n_samples=1000,n_features=2,n_classes=2,n_clusters_per_class=1, n_redundant=0, )
X
```

```python
y
```

```python
df=pd.DataFrame(X)
df[0],df[1]
```

```python
sns.scatterplot(x=df[0],y=df[1],hue=y)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.25, random_state=42)
```

```python
from sklearn.svm import SVC
```

```python
svc = SVC(kernel='linear')
# as we can see the points are linearly saperable we can use linear kernel to saperate datapoints
```

```python
svc.fit(X_train,y_train)
```

```python
y_pred=svc.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix,classification_report
cm=confusion_matrix(y_test,y_pred)
class_rep=classification_report(y_test, y_pred)
print("confusion matrics :\n",cm)
print("classification report:\n",class_rep)
```

```python
# Create grid (canvas)
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1

xx, yy = np.meshgrid(
    np.linspace(x_min, x_max, 200),
    np.linspace(y_min, y_max, 200)
)

# Predict on grid points
Z = svc.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

# Plot decision boundary
plt.contourf(xx, yy, Z, alpha=0.3)

# Plot original data using seaborn style
sns.scatterplot(x=X[:, 0], y=X[:, 1], hue=y)

plt.title("SVM Decision Boundary")
plt.show()
```

## now we will see how different kernels classify the following dataset 


```python
X,y=make_classification(n_samples=1000,n_features=2,n_classes=2,n_clusters_per_class=2, n_redundant=0, )
df=pd.DataFrame(X)
sns.scatterplot(x=df[0],y=df[1],hue=y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.25, random_state=42)
```

## linear kernel

```python
linear = SVC(kernel='linear')
linear.fit(X_train,y_train)
y_pred1=linear.predict(X_test)
cm1=confusion_matrix(y_test,y_pred1)
class_rep1=classification_report(y_test, y_pred1)
print("confusion matrics :\n",cm1)
print("classification report:\n",class_rep1)
```

```python
# Create grid (canvas)
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1

xx, yy = np.meshgrid(
    np.linspace(x_min, x_max, 200),
    np.linspace(y_min, y_max, 200)
)

# Predict on grid points
Z = svc.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

# Plot decision boundary
plt.contourf(xx, yy, Z, alpha=0.3)

# Plot original data using seaborn style
sns.scatterplot(x=X[:, 0], y=X[:, 1], hue=y)

plt.title("SVM Decision Boundary")
plt.show()
```

## polynomial kernel

```python
poly = SVC(kernel='poly')
poly.fit(X_train,y_train)
y_pred2=poly.predict(X_test)
cm2=confusion_matrix(y_test,y_pred2)
class_rep2=classification_report(y_test, y_pred2)
print("confusion matrics :\n",cm2)
print("classification report:\n",class_rep2)
```

## rbf kernel

```python
rgb = SVC(kernel='rbf')
rgb.fit(X_train,y_train)
y_pred3=rgb.predict(X_test)
cm3=confusion_matrix(y_test,y_pred3)
class_rep3=classification_report(y_test, y_pred3)
print("confusion matrics :\n",cm3)
print("classification report:\n",class_rep3)
```

## sigmoid function

```python
sigmoid = SVC(kernel="sigmoid")
sigmoid.fit(X_train,y_train)
y_pred4=sigmoid.predict(X_test)
cm4=confusion_matrix(y_test,y_pred4)
class_rep4=classification_report(y_test, y_pred4)
print("confusion matrics :\n",cm4)
print("classification report:\n",class_rep4)
```

# Hyperparameter Tuning

- as we have seen how different kernels saperates the point differently with different accuracy so 
- now we can select best kernel using hyperparameter tuning

```python
from sklearn.model_selection import GridSearchCV
```

```python
params_grid={
    'C':[0.1,1,10,100,1000],
    'gamma':[1,0.1,0.01,0.001,0.0001],
    'kernel':['rbf']
}
```

```python
grid=GridSearchCV(SVC(),param_grid=params_grid,refit=True,cv=5,verbose=3)
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
y_pred5=grid.predict(X_test)
cm5=confusion_matrix(y_test,y_pred5)
class_rep5=classification_report(y_test, y_pred5)
print("confusion matrics :\n",cm5)
print("classification report:\n",class_rep5)
```
